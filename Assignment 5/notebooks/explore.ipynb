```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number, col
```

```python
window_spec = Window.partitionBy("customer_id").orderBy(
    col("order_purchase_timestamp").asc()
)

print("Window specification created!")
```

```python
window_spec = Window.partitionBy("customer_id").orderBy(
    col("order_purchase_timestamp").asc()
)

print("Window specification created!")
```

```python
ranked_orders_df = orders_df.withColumn(
    "order_rank",
    row_number().over(window_spec)
)

ranked_orders_df.select(
    "customer_id",
    "order_id",
    "order_purchase_timestamp",
    "order_rank"
).show(10, truncate=False)
```

```python
orders_per_customer = orders_df.groupBy("customer_id").count()

orders_per_customer.orderBy(
    col("count").desc()
).show(10)
```

```python
from pyspark.sql.functions import rank, dense_rank
```

```python
from pyspark.sql import Row

products = [
    Row(product="Laptop", price=1000),
    Row(product="Phone", price=800),
    Row(product="Tablet", price=800),
    Row(product="Mouse", price=200)
]

products_df = spark.createDataFrame(products)

price_window = Window.orderBy(col("price").desc())

ranked_products = products_df.withColumn(
    "rank",
    rank().over(price_window)
).withColumn(
    "dense_rank",
    dense_rank().over(price_window)
)

ranked_products.show()
```

```python
latest_window = Window.partitionBy("customer_id").orderBy(
    col("order_purchase_timestamp").desc()
)

latest_orders_df = orders_df.withColumn(
    "row_num",
    row_number().over(latest_window)
)

latest_orders_df.filter(
    col("row_num") == 1
).select(
    "customer_id",
    "order_id",
    "order_purchase_timestamp"
).show(10, truncate=False)
```
