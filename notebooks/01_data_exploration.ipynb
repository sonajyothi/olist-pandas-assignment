### Missing approval dates

There are 160 missing values in `order_approved_at`.

Most are associated with cancelled orders (141). There are also 14 delivered orders and 5 created orders with missing approval dates.

The missing values were kept as `NaT` rather than being filled with an estimated date, because an approval timestamp cannot be reliably inferred from the available columns.
